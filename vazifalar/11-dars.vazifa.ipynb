```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler


df = pd.read_csv("Procurement KPI Analysis Dataset.csv")


numeric_columns = ['Quantity', 'Unit_Price', 'Negotiated_Price', 'Defective_Units']


df[numeric_columns] = df[numeric_columns].fillna(0)


min_max_scaler = MinMaxScaler()
min_max_scaled = min_max_scaler.fit_transform(df[numeric_columns])
min_max_df = pd.DataFrame(min_max_scaled, columns=[f"{col}_minmax" for col in numeric_columns])


standard_scaler = StandardScaler()
standard_scaled = standard_scaler.fit_transform(df[numeric_columns])
standard_df = pd.DataFrame(standard_scaled, columns=[f"{col}_zscore" for col in numeric_columns])


result_df = pd.concat([df[['PO_ID']], min_max_df, standard_df], axis=1)


print(result_df.head())
```
