```python
import pandas as pd
```

```python
df=pd.read_csv('Supplement_Sales_Weekly_Expanded.csv')
```

```python
df.info()
```

```python
for col in df.columns:
    if df[col].isnull().sum()>0:
        if df[col].dtype=='object':
            df[col].fillna(df[col].mode()[0],inplace=True)
        else:
            df[col].fillna(df[col].mean(),inplace=True)
```
