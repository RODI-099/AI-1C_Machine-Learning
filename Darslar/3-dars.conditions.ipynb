```python
# Agar x=1 ga teng bolsa ok degan soni ekranga chiqaring x=-1 unda manfiy son degan ekranga chiqaring aks xolda xato degan xabarni chiqaring
x=int(input('Sonni kiriting:'))
if x==1:
    print('Ok')
elif x==-1:
    print('Manfiy son')
else:
    print('Xato')
```

```python
# Agar x=1 ga teng bolsa ok degan soni ekranga chiqaring x=-1 unda manfiy son degan ekranga chiqaring aks xolda xato degan xabarni chiqaring
x=int(input('Sonni kiriting:'))
if x==1:
    print('Ok')
elif x==-1 and x==0:
    print('Manfiy son yoki nolga teng')

else:
    print('Xato')
```
